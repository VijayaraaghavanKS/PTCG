```python
from pathlib import Path
from collections import Counter
import base64, hashlib, importlib, os, shutil, sys, tarfile, tempfile

OUTPUT_DIR = Path(os.environ.get("PTCG_WORK_DIR", "/kaggle/working"))
INPUT_ROOT = Path(os.environ.get("PTCG_INPUT_ROOT", "/kaggle/input"))
COMPETITION_ROOT = Path(os.environ.get(
    "PTCG_COMPETITION_ROOT",
    "/kaggle/input/competitions/pokemon-tcg-ai-battle",
))
MAX_SUBMISSION_BYTES = 100 * 1024 * 1024
ASSET_SHA256 = '40dce050fc411a2845b3dcd364fdd932ae3a856720a9f5f022278c59dd6e3a72'
ASSET_B64 = ''

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
SUBMISSION = OUTPUT_DIR / "submission.tar.gz"
SUBMISSION.unlink(missing_ok=True)
TEMP_ROOT = Path("/kaggle/temp") if Path("/kaggle/temp").exists() else Path(tempfile.gettempdir())
BUILD = Path(tempfile.mkdtemp(prefix="grimmsnarl_build_", dir=TEMP_ROOT))
ASSET = BUILD.parent / f"{BUILD.name}_assets.tar.gz"
ASSET.write_bytes(base64.b64decode(ASSET_B64))
assert hashlib.sha256(ASSET.read_bytes()).hexdigest() == ASSET_SHA256
print("Embedded policy assets: PASS")
```

```python
def safe_extract(archive: Path, destination: Path) -> None:
    with tarfile.open(archive, "r:gz") as tf:
        root = destination.resolve()
        for member in tf.getmembers():
            target = (destination / member.name).resolve()
            assert target == root or root in target.parents, member.name
            assert not member.issym() and not member.islnk(), member.name
        try:
            tf.extractall(destination, filter="data")
        except TypeError:
            tf.extractall(destination)


def valid_cg_dir(path: Path) -> bool:
    return all((path / name).is_file() for name in (
        "libcg.so", "api.py", "game.py", "sim.py", "utils.py"
    ))


def find_official_cg() -> Path:
    candidates = []
    for root in (COMPETITION_ROOT, INPUT_ROOT):
        if not root.exists():
            continue
        for lib in root.rglob("libcg.so"):
            cg_dir = lib.parent
            if valid_cg_dir(cg_dir):
                text = str(cg_dir).lower()
                score = 3 * ("pokemon-tcg-ai-battle" in text) + 2 * ("sample_submission" in text)
                candidates.append((-score, len(text), cg_dir))
    if not candidates:
        raise FileNotFoundError(
            "Official competition runtime not found. Add the Pokémon TCG AI Battle competition input."
        )
    return sorted(candidates)[0][2]

safe_extract(ASSET, BUILD)
CG_SOURCE = find_official_cg()
shutil.copytree(CG_SOURCE, BUILD / "cg", dirs_exist_ok=True)
(BUILD / "cg" / "__init__.py").touch(exist_ok=True)
print("Official runtime copied:", CG_SOURCE)
```

```python
required = {
    "main.py", "deck.csv", "EN_Card_Data.csv",
    "models/feature_schema.pkl.gz", "models/policy_ensemble.bin.gz", "residual_guard.py", "advisor_guard.py", "tactical_guard.py", "development_guard.py", "robustness_guard.py",
    "cg/__init__.py", "cg/api.py", "cg/game.py", "cg/sim.py", "cg/utils.py", "cg/libcg.so",
}
found = {path.relative_to(BUILD).as_posix() for path in BUILD.rglob("*") if path.is_file()}
missing = sorted(required - found)
assert not missing, missing

deck = [int(value) for value in (BUILD / "deck.csv").read_text().split()]
assert len(deck) == 60

# Import the packaged entry point without importing cg.game in the notebook.
sys.path.insert(0, str(BUILD))
old_cwd = Path.cwd()
try:
    os.chdir(BUILD)
    sys.modules.pop("main", None)
    submission_main = importlib.import_module("main")
finally:
    os.chdir(old_cwd)

assert deck == list(submission_main.EXPECTED_DECK)
assert deck == list(submission_main.DECK)
assert deck == list(submission_main.agent({}))
print("Fixed 60-card deck: PASS")
print("Package structure: PASS")
```

```python
# Package directly into Kaggle's persistent output directory.
for cache in BUILD.rglob("__pycache__"):
    shutil.rmtree(cache, ignore_errors=True)

TEMP_SUBMISSION = OUTPUT_DIR / "submission.tar.gz.tmp"
TEMP_SUBMISSION.unlink(missing_ok=True)
with tarfile.open(TEMP_SUBMISSION, "w:gz", compresslevel=9) as tf:
    for path in sorted(BUILD.rglob("*")):
        if path.is_file():
            tf.add(path, arcname=path.relative_to(BUILD).as_posix())
TEMP_SUBMISSION.replace(SUBMISSION)

with tarfile.open(SUBMISSION, "r:gz") as tf:
    members = [member.name for member in tf.getmembers() if member.isfile()]
    archived_deck = [int(value) for value in tf.extractfile("deck.csv").read().decode().split()]

assert required.issubset(set(members))
assert archived_deck == deck
assert SUBMISSION.is_file() and SUBMISSION.stat().st_size > 0
assert SUBMISSION.stat().st_size < MAX_SUBMISSION_BYTES
assert not [name for name in members if name.startswith("/") or ".." in Path(name).parts]

shutil.rmtree(BUILD, ignore_errors=True)
ASSET.unlink(missing_ok=True)

output_files = sorted(path for path in OUTPUT_DIR.iterdir() if path.is_file())
assert SUBMISSION in output_files
print("Submission ready:", SUBMISSION)
print("Size (MB):", round(SUBMISSION.stat().st_size / 1024**2, 2))
print("Notebook outputs:", [path.name for path in output_files])
print("Submission output: PASS")
```
